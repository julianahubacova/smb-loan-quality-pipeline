```python
# ── EXPORT TABLES FOR POWER BI ──────────────────────────────
# Export Gold and DQ tables to CSV for Power BI consumption

import pandas as pd

# Export gold_loans_by_province
df_province = spark.table("gold_loans_by_province").toPandas()
print("Loans by Province:")
print(df_province)

# Export gold_loans_by_industry
df_industry = spark.table("gold_loans_by_industry").toPandas()
print("\nLoans by Industry:")
print(df_industry)

# Export dq_results
df_dq = spark.table("dq_results").toPandas()
print("\nDQ Results:")
print(df_dq)
```

```python
# Save CSVs locally so we can download them
df_province.to_csv("/tmp/gold_loans_by_province.csv", index=False)
df_industry.to_csv("/tmp/gold_loans_by_industry.csv", index=False)
df_dq.to_csv("/tmp/dq_results.csv", index=False)

print("CSVs saved! Now download them using the code below:")

# Display download links
for filename in ["gold_loans_by_province.csv", "gold_loans_by_industry.csv", "dq_results.csv"]:
    print(f"/tmp/{filename}")
```
